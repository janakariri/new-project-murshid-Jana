```python
# Deliberate, caught errors. These are examples to diagnose, not broken setup.
import json


def validate_public_question(question):
    if not isinstance(question, str) or not question.strip():
        raise ValueError('Enter a non-empty question string.')
    if len(question) > 500:
        raise ValueError('Use at most 500 characters.')
    return question.strip()


def run_error_examples():
    cases = [(None, 'Enter a non-empty question string.'),
             ('   ', 'Enter a non-empty question string.'),
             ('x' * 501, 'Use at most 500 characters.')]
    rows = []
    for value, expected in cases:
        try:
            validate_public_question(value)
        except ValueError as error:
            assert str(error) == expected
            rows.append({'error_type': type(error).__name__, 'message': str(error), 'expected_rejection': True})
        else:
            raise AssertionError('Invalid input was accepted.')
    assert validate_public_question(' How can I check my application status? ') == 'How can I check my application status?'
    result = {'intentional_caught_errors': rows, 'corrected_input': 'accepted', 'checks': 'passed'}
    print(json.dumps(result, indent=2))
    return result
```

```python
error_result = run_error_examples()
```

## Homework explanation

**Predictions and results.**

| Input | Prediction | Result |
|---|---|---|
| `None` | rejected: not a string | `ValueError: Enter a non-empty question string.` |
| `'   '` (blank) | rejected: empty after stripping | `ValueError: Enter a non-empty question string.` |
| `'x' * 501` | rejected: too long | `ValueError: Use at most 500 characters.` |
| `' How can I check my application status? '` | accepted | returned the stripped question |

**Why a caught rejection can be a successful test.** The input is invalid on purpose, and the function is supposed to reject it with a clear message. The test passes when the right error is raised for the right reason, and fails if invalid input is accepted. The errors are caught with `try/except`, so the notebook does not crash.

**How the input was corrected.** A non-empty question of at most 500 characters passes validation, and surrounding spaces are removed.

**Limitation.** This is a standalone validation exercise, not an automatic check inside every model function. Three invalid cases do not cover every possible bad input.
